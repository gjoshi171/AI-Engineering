```python
from langsmith import Client
```

### Attach Feedback to a specific trace in LangSmith

```python
client = Client()
```

```python
RUN_ID = "019faf60-3d63-7ab2-9ad7-3d0bca04f801"
```

```python
client.create_feedback(
    run_id = RUN_ID,
    key = "feedback",
    score=0,
    feedback_source_type = "api"
)
```

```python
client.create_feedback(
    run_id = RUN_ID,
    key = "comment",
    value="This is a long form free text feedback left by GAurav",
    feedback_source_type = "api"
)
```

```python
client.create_feedback(
    run_id = RUN_ID,
    key = "comment_with_score",
    value="This is a long form free text feedback left by GAurav",
    score=1,
    feedback_source_type = "api"
)
```
